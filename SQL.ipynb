```python
import pandas as pd
import sqlite3

df = pd.read_csv(r"C:\Users\ammar\OneDrive\Desktop\Sample - Superstore.csv" , encoding = 'latin1' )

#  Create a new SQLite database
conn = sqlite3.connect("superstore.db")

# Store the data in a table
df.to_sql("superstore", conn, if_exists="replace", index=False)

#  Verify the data
print("Rows inserted:", pd.read_sql("SELECT COUNT(*) AS total FROM superstore", conn))
```

```python
query_a = """
SELECT Region, Category, SUM(Sales) AS Total_Sales
FROM superstore
WHERE Sales > 0
GROUP BY Region, Category
ORDER BY Total_Sales DESC;
"""
pd.read_sql(query_a, conn)
```

```python
# Create a small lookup table for region manager names
region_managers = pd.DataFrame({
    "Region": ["East", "West", "Central", "South"],
    "Manager": ["Alice", "Bob", "Charlie", "Diana"]
})

region_managers.to_sql("region_managers", conn, if_exists="replace", index=False)

query_b = """
SELECT s.Region, s.Category, SUM(s.Sales) AS Total_Sales, r.Manager
FROM superstore s
LEFT JOIN region_managers r
ON s.Region = r.Region
GROUP BY s.Region;
"""
pd.read_sql(query_b, conn)
```

```python
query_c = """
SELECT * FROM superstore
WHERE Sales > (SELECT AVG(Sales) FROM superstore);
"""
pd.read_sql(query_c, conn)
```

```python
query_c = """
SELECT * FROM superstore
WHERE Sales > (SELECT AVG(Sales) FROM superstore);
"""
pd.read_sql(query_c, conn)
```

```python
query_d = """
SELECT Category, SUM(Sales) AS Total_Sales, AVG(Profit) AS Avg_Profit
FROM superstore
GROUP BY Category;
"""
pd.read_sql(query_d, conn)
```

```python
conn.execute("""
CREATE VIEW IF NOT EXISTS sales_summary AS
SELECT Region, Category, SUM(Sales) AS Total_Sales, SUM(Profit) AS Total_Profit
FROM superstore
GROUP BY Region, Category;
""")

# Check the view
pd.read_sql("SELECT * FROM sales_summary", conn)
```
